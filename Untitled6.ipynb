```python

import numpy as np
import matplotlib.pyplot as plt

# Power system parameters
demand = 450
capacity = np.array([250, 200, 180])
loss_factor = np.array([0.0008, 0.0012, 0.0005])

# PCA parameters
size = 10
iterations = 100

# Generate a feasible power distribution
def distribute(weights):
    weights = np.maximum(weights, 0.001)

    low, high = 0, 10000
    for _ in range(50):
        mid = (low + high) / 2
        power = np.minimum(mid * weights, capacity)

        if power.sum() < demand:
            low = mid
        else:
            high = mid

    return np.minimum(high * weights, capacity)

# Calculate power loss (fitness function)
def fitness(solution):
    return np.sum(loss_factor * solution**2)

# Initialize 10x10 cells
grid = np.array([
    [distribute(np.random.rand(3)) for j in range(size)]
    for i in range(size)
])

best_history = []

# Parallel Cellular Algorithm
for iteration in range(iterations):

    new_grid = grid.copy()

    for i in range(size):
        for j in range(size):

            # Find best neighboring cell
            neighbors = grid[
                max(0, i-1):min(size, i+2),
                max(0, j-1):min(size, j+2)
            ]

            best_neighbor = min(
                neighbors.reshape(-1, 3),
                key=fitness
            )

            # Update cell using neighbor's solution
            new_weights = best_neighbor + np.random.normal(0, 5, 3)
            new_grid[i, j] = distribute(new_weights)

    grid = new_grid

    # Find best solution
    best = min(grid.reshape(-1, 3), key=fitness)
    best_history.append(fitness(best))

# Display results
best = min(grid.reshape(-1, 3), key=fitness)

print("Optimal Power Distribution:")
for i, power in enumerate(best):
    print(f"Area {i+1}: {power:.2f} units")

print("Total Power:", round(best.sum(), 2))
print("Minimum Power Loss:", round(fitness(best), 2))

# Plot convergence graph
plt.plot(best_history)
plt.xlabel("Iterations")
plt.ylabel("Power Loss")
plt.title("PCA - Power System Optimization")
plt.grid()
plt.show()

# Plot power distribution
plt.bar(["Area A", "Area B", "Area C"], best)
plt.xlabel("Areas")
plt.ylabel("Power Distribution")
plt.title("Optimal Electricity Distribution")
plt.show()
```
